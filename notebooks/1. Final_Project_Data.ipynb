# 1. Project Overview and Data Sources

In this notebook, I prepare the data for my final project. The goal is to clean and combine multiple datasets into a relational database. 

I use four main data sources for this project:
1. **TCGA Clinical Data**: 15 files of different cancer types from The Cancer Genome Atlas (TCGA). I use this data to create the clinical patients database.

3. **Pharmacology Protocols**: A database of current medical treatments. This is based on international clinical guidelines. Sources include:
   * [NCCN (National Comprehensive Cancer Network)](https://www.nccn.org/)
   * [ESMO (European Society for Medical Oncology)](https://www.esmo.org/)
   * [ASCO (American Society of Clinical Oncology)](https://www.asco.org/)
   * [FDA (U.S. Food and Drug Administration)](https://www.fda.gov/)
4. **NHANES Lifestyle and Survival**: A database combining public health surveys (NHANES) to link cancer diagnosis, lifestyle habits, and survival rates.

# 4. Data Cleaning

After combining the datasets, I checked for missing values using `df_onco.isnull().sum()`. I quickly realized that real-world medical data has a lot of missing information. For example, some columns had over 6,000 nulls.

Because of this, I performed Feature Selection and Missing Data Imputation. I decided to keep only the most important features needed for my Streamlit dashboard (patient demographics, diagnostic markers, and survival statistics).


```python
pd.set_option('display.max_rows', None)
df_onco.isnull().sum()
```

### 4.1. Feature engineering and cleaning decisions:

* **Core Identifiers (0 Missing Values):** Columns like `Study ID`, `Patient ID`, and `Cancer Type` are perfectly populated. I use these as my primary keys to link tables.
* **Medical Staging (~1,800 to 2,500 Missing Values):** Tumor stage columns had many nulls. Instead of dropping these rows (which would drastically reduce my dataset), I filled them with 'Not Applicable / Unknown'. This is because some cancers, like Glioblastoma, do not follow standard staging protocols.
* **Demographics (~450 to 550 Missing Values):** Variables like `Sex` and `Diagnosis Age` had a few missing values. Since `Diagnosis Age` is a crucial input for my app, I filled the missing cells with the dataset's median to keep the data complete without creating outliers.
* **Survival Outcomes (~77 to 115 Missing Values):** For critical metrics like `Overall Survival (Months)`, I strictly dropped the rows with missing data using `.dropna()`. I need this specific data to be 100% accurate for the dashboard's calculations.
* **Unreliable Variables (>3,000 Missing Values):** Columns with massive amounts of missing data (like `Patient Weight` or `Disease Free Months`) were completely removed during feature selection. Trying to impute thousands of rows would introduce artificial noise into the database.

To make sure my Streamlit app runs smoothly and my models stay accurate, I created a clear data cleaning process. Real-world medical data from different hospitals often has missing values. Instead of simply deleting rows with missing data (which would shrink my dataset too much). I handled them based on their data type:

* **Categorical Data (Text):** I filled empty fields in clinical categories (like tumor stages or therapy types) with 'Not Applicable / Unknown'. This keeps the patient history traceable and works perfectly with the dropdown filters in my app.
* **Numerical Data (Biomarkers):** For missing genomic numbers (like TMB or MSIsensor scores), I filled the blanks with the median value. This fixes the missing data without creating weird outliers that could ruin the statistics.
* **Column Names (Formatting):** I changed all 64 column names to `lowercase_snake_case`, removing spaces and special characters. This makes the code much cleaner and prevents errors when moving the data into SQL databases.

### 4.1.1. Creating a copy to safeguard original data

```python
df_curated = df_onco.copy()
```

### 4.1.2. Programmatic Column Standardization (Lowercase Snake_Case Schema)
Removing spaces, brackets, hyphens, and standardizing characters for clean relational integration

```python
df_curated.columns = (
    df_curated.columns.str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
    .str.replace("-", "_", regex=False)
    .str.replace(",", "", regex=False)
)
```

### 4.1.3. Categorical imputation

```python
categorical_impute_cols = [
    "neoplasm_disease_stage_american_joint_committee_on_cancer_code",
    "american_joint_committee_on_cancer_publication_version_type",
    "ethnicity_category",
    "neoplasm_histologic_grade",
    "neoadjuvant_therapy_type_administered_prior_to_resection_text",
    "american_joint_committee_on_cancer_metastasis_stage_code",
    "neoplasm_disease_lymph_node_stage_american_joint_committee_on_cancer_code",
    "american_joint_committee_on_cancer_tumor_stage_code",
    "person_neoplasm_cancer_status",
    "race_category",
    "radiation_therapy",
    "subtype",
    "disease_free_status",
]

for col in categorical_impute_cols:
    if col in df_curated.columns:
        df_curated[col] = df_curated[col].fillna("Not applicable / Unknown")
```

### 4.1.4. Numerical Imputation (Median)

```python
numerical_biomarkers = [
    "diagnosis_age",
    "aneuploidy_score",
    "buffa_hypoxia_score",
    "winter_hypoxia_score",
    "ragnum_hypoxia_score",
    "fraction_genome_altered",
    "msi_mantis_score",
    "msisensor_score",
    "mutation_count",
    "tmb_nonsynonymous",
    "tumor_break_load",
]

for col in numerical_biomarkers:
    if col in df_curated.columns:
        median_value = df_curated[col].median()
        # In case the entire column is unpopulated, fallback safety to 0
        if pd.isna(median_value):
            median_value = 0
        df_curated[col] = df_curated[col].fillna(median_value)
```

### 4.1.5. Drop rows with missing core target metrics

```python
core_targets = ["overall_survival_status", "overall_survival_months"]
df_curated = df_curated.dropna(subset=[col for col in core_targets if col in df_curated.columns])
```

# 6. NHANES Lifestyle and Survival Data

I merge multiple SAS files from the NHANES public health survey. This links cancer diagnoses with lifestyle habits and mortality records.

### 6.1. Loading NHANES files and Loading mortality fixed-width file

```python

df_demo = pd.read_sas("../data/raw/DEMO_H_2013-2014.xpt")
df_mcq = pd.read_sas("../data/raw/MCQ_H_2013-2014.xpt")
df_paq = pd.read_sas("../data/raw/PAQ_H_2013-2014.xpt")
df_smoking = pd.read_sas("../data/raw/SMOKING_2013-2014.xpt")

widths = [14, 1, 1, 3, 1, 1, 1, 4, 8, 8, 3, 3]
names = ['SEQN', 'ELIGSTAT', 'MORTSTAT', 'UCOD_LEADING', 'DIABETES', 'HYPERTEN', 
         'DODQTR', 'DODYEAR', 'WGT_NEW', 'SA_WGT_NEW', 'PERMTH_INT', 'PERMTH_EXM']
df_mort = pd.read_fwf("../data/raw/NHANES_2013_2014_MORT_2019_PUBLIC.dat", widths=widths, names=names)
```

### 6.2. Merging databases

```python
df_nhanes_master = pd.merge(df_demo, df_mcq, on='SEQN', how='inner')
df_nhanes_master = pd.merge(df_nhanes_master, df_paq, on='SEQN', how='inner')
df_nhanes_master = pd.merge(df_nhanes_master, df_smoking, on='SEQN', how='inner')
df_nhanes_master = pd.merge(df_nhanes_master, df_mort, on='SEQN', how='left')
```

### 6.3. Filtering for patients with a cancer diagnosis

```python
df_nhanes_analytics = df_nhanes_master[df_nhanes_master['MCQ220'] == 1.0].copy()

rename_dict = {
    'RIDAGEYR': 'age',
    'RIAGENDR': 'gender',
    'MCQ220': 'cancer_diagnosis',
    'PAD680': 'sedentary_minutes_day',
    'PAQ650': 'vigorous_recreation',
    'SMQ020': 'smoked_100_cigs',
    'SMQ040': 'current_smoking_status',
    'PERMTH_INT': 'survival_months',
    'MORTSTAT': 'mortality_status'
}
df_nhanes_analytics.rename(columns=rename_dict, inplace=True)
```

### 6.4. Pushing the final analytics table to Neon

```python
df_nhanes_analytics.to_sql('nhanes_analytics_data', con=engine, if_exists='replace', index=False)
```
